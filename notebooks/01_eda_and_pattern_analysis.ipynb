**Objective:** Explore the dynamic drivers of student mess attendance—day of week, academic exams, menu popularity, holidays, and weather conditions—to quantify the baseline food waste caused by status-quo fixed cooking policies.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid', palette='muted')
plt.rcParams['figure.figsize'] = (12, 6)

# Load historical mess operations log
df = pd.read_csv('../data/mess_attendance_raw.csv')
print(f'Total meal sessions loaded: {len(df)}')
df.head()
```

### 1. Attendance Distribution by Meal Session
In typical hostel setups, students often skip breakfast (due to late-night study or rush to 8:30 AM lectures), while lunch and dinner represent the major food preparation volumes.

```python
fig, ax = plt.subplots(1, 2, figsize=(14, 5))
sns.boxplot(data=df, x='meal_type', y='actual_attendance', ax=ax[0], palette='Set2')
ax[0].set_title('Attendance Distribution by Meal Type')
ax[0].set_ylabel('Student Headcount')

sns.histplot(data=df, x='actual_attendance', hue='meal_type', kde=True, ax=ax[1], palette='Set2')
ax[1].set_title('Turnout Density Across Meals')
plt.tight_layout()
plt.show()
```

### 2. Temporal Day-of-the-Week Attrition & Weekend Attrition
Friday dinner and weekends exhibit massive student drop-offs as hostellers visit local eateries, order food online, or travel home during long weekends.

```python
dow_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
plt.figure(figsize=(10, 5))
sns.barplot(data=df, x='day_name', y='actual_attendance', order=dow_order, palette='crest')
plt.title('Average Student Attendance Across Days of the Week')
plt.ylabel('Average Turnout')
plt.show()
```

### 3. Menu Popularity Elasticity
Special feasts (Paneer Butter Masala, Biryani, Chole Bhature) draw over 85-95% attendance, whereas simpler menus (Lauki, Tinda, Khichdi) drop below 50%.

```python
plt.figure(figsize=(10, 5))
sns.regplot(data=df.sample(500, random_state=42), x='menu_popularity', y='actual_attendance',
            scatter_kws={'alpha':0.4, 'color':'teal'}, line_kws={'color':'crimson'})
plt.title('Menu Popularity Score vs Student Headcount')
plt.xlabel('Menu Popularity Rating (1-10)')
plt.ylabel('Actual Attendance')
plt.show()
```

### 4. Status Quo Baseline Waste Quantification
Cooking for a fixed 85% capacity (850 students) causes systematic overcooking whenever attendance is low.

```python
avg_waste_kg = df['baseline_waste_kg'].mean()
total_waste_kg = df['baseline_waste_kg'].sum()
total_cost_inr = df['baseline_waste_cost_inr'].sum()

print(f'--- Baseline Cooking Waste Summary ---')
print(f'Average food wasted per meal: {avg_waste_kg:.1f} kg')
print(f'Total food wasted across {len(df)} meals: {total_waste_kg:,.0f} kg ({total_waste_kg/1000:.1f} tonnes)')
print(f'Total direct food expenditure wasted: ₹ {total_cost_inr:,.0f}')
```
